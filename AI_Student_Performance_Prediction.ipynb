```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pickle

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("✅ Libraries imported successfully!")
```

```python
# Create sample student performance dataset
np.random.seed(42)

n = 120

study_hours = np.random.randint(1, 11, n)
attendance_percent = np.random.randint(55, 101, n)
assignment_score = np.random.randint(40, 101, n)
previous_exam_score = np.random.randint(40, 101, n)
sleep_hours = np.round(np.random.uniform(5, 9, n), 1)

score = (
    study_hours * 5
    + attendance_percent * 0.25
    + assignment_score * 0.20
    + previous_exam_score * 0.30
    + sleep_hours * 1.5
    + np.random.normal(0, 8, n)
)

performance = np.where(
    score >= 100, "High",
    np.where(score >= 80, "Medium", "Low")
)

student_df = pd.DataFrame({
    "study_hours": study_hours,
    "attendance_percent": attendance_percent,
    "assignment_score": assignment_score,
    "previous_exam_score": previous_exam_score,
    "sleep_hours": sleep_hours,
    "performance": performance
})

print("Dataset Shape:", student_df.shape)
display(student_df.head())
```

```python
print("Missing Values:")
print(student_df.isnull().sum())

print("\nPerformance Distribution:")
print(student_df["performance"].value_counts())

print("\nDataset Information:")
student_df.info()
```

```python
# Exploratory data analysis
plt.figure(figsize=(7, 5))
sns.countplot(data=student_df, x="performance")
plt.title("Student Performance Distribution")
plt.xlabel("Performance Level")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Feature selection
X = student_df[
    [
        "study_hours",
        "attendance_percent",
        "assignment_score",
        "previous_exam_score",
        "sleep_hours"
    ]
]

y = student_df["performance"]

print("Features:")
display(X.head())

print("\nTarget:")
display(y.head())
```

```python
# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
# Train Random Forest model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("✅ Random Forest model trained successfully!")
```

```python
# Evaluate the model
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy * 100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=["Low", "Medium", "High"])

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["Low", "Medium", "High"],
    yticklabels=["Low", "Medium", "High"]
)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# Feature importance
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
}).sort_values("Importance", ascending=False)

display(importance)

plt.figure(figsize=(8, 5))
sns.barplot(data=importance, x="Importance", y="Feature")
plt.title("Feature Importance")
plt.show()
```

```python
# Interactive student performance prediction

def predict_student_performance(
    study_hours,
    attendance_percent,
    assignment_score,
    previous_exam_score,
    sleep_hours
):
    student = pd.DataFrame({
        "study_hours": [study_hours],
        "attendance_percent": [attendance_percent],
        "assignment_score": [assignment_score],
        "previous_exam_score": [previous_exam_score],
        "sleep_hours": [sleep_hours]
    })

    prediction = model.predict(student)[0]

    print("🎓 Student Performance Prediction")
    print("=" * 40)
    print(f"Study Hours: {study_hours}")
    print(f"Attendance: {attendance_percent}%")
    print(f"Assignment Score: {assignment_score}")
    print(f"Previous Exam Score: {previous_exam_score}")
    print(f"Sleep Hours: {sleep_hours}")
    print(f"\n🤖 Predicted Performance: {prediction}")

    return prediction

predict_student_performance(
    study_hours=8,
    attendance_percent=90,
    assignment_score=88,
    previous_exam_score=85,
    sleep_hours=7
)
```

```python
# Test multiple students
test_students = [
    [2, 60, 55, 50, 5.5],
    [5, 75, 70, 68, 6.5],
    [9, 95, 92, 90, 8]
]

for student in test_students:
    print("-" * 50)
    predict_student_performance(*student)
```

```python
# Save dataset and trained model
student_df.to_csv("student_performance_dataset.csv", index=False)

with open("student_performance_model.pkl", "wb") as file:
    pickle.dump(model, file)

print("✅ Dataset saved: student_performance_dataset.csv")
print("✅ Model saved: student_performance_model.pkl")
```
